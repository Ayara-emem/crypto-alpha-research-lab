## Interpretation

The descriptive statistics indicate that Bitcoin prices span a wide range over the sample period, reflecting substantial appreciation as well as periods of severe market decline. The large difference between the minimum and maximum prices suggests the presence of multiple market regimes rather than a single stable price process.

Because price levels evolve over time and generally exhibit non-stationary behavior, these statistics should be interpreted as descriptive rather than predictive. Subsequent analysis will therefore focus primarily on return series, which provide a more suitable basis for statistical modeling and hypothesis testing.
