# Notebook 05: Volume Analysis

### Research Question

How has Bitcoin trading volume evolved over time, and what information does trading activity provide about market participation, liquidity, and investor behavior?

### Research Objectives

- Examine the time-series behavior of Bitcoin trading volume.
- Identify periods of unusually high or low market activity.
- Assess whether trading volume exhibits persistent trends.
- Relate volume dynamics to market liquidity and investor participation.

### Why This Matters

Trading volume is a fundamental measure of market activity and liquidity. In financial economics, periods of elevated trading volume often coincide with increased information arrival, stronger price discovery, and heightened investor participation. Understanding volume dynamics provides valuable context for interpreting market volatility and future signal development.

```python
TICKER = "BTC-USD"

START_DATE = "2020-01-01"
END_DATE = "2026-06-20"

INTERVAL = "1d"

AUTO_ADJUST = True

REFRESH_CACHE = True
```

```python
import matplotlib.pyplot as plt

from crypto_alpha_lab.data.storage import load_prices
```

```python
btc = load_prices(
    ticker=TICKER,
    start=START_DATE,
    end=END_DATE,
    interval=INTERVAL,
    auto_adjust=AUTO_ADJUST,
    refresh=REFRESH_CACHE,
)
```

```python
btc["Volume"].head()
```

```python
btc["Volume"].describe()
```

```python
plt.figure(figsize=(14,6))

plt.plot(
    btc.index,
    btc["Volume"],
    linewidth=1,
)

plt.title("Bitcoin Daily Trading Volume")
plt.xlabel("Date")
plt.ylabel("Trading Volume")

plt.grid(True)

plt.show()
```

### Volume Interpretation

Bitcoin trading volume exhibits pronounced time variation with several distinct episodes of intensified market activity. The most prominent spike occurs in early **2021**, where daily trading volume reaches approximately **3.5 × 10¹¹**, representing the highest level observed during the sample period.

Following this peak, trading activity moderates throughout **2022–2023**, indicating a period of relatively lower market participation. From **2024 onward**, volume becomes increasingly volatile with frequent large spikes, suggesting renewed investor activity, improved liquidity, and heightened information flow. Overall, the volume series demonstrates that Bitcoin market participation is dynamic and closely linked to changing market conditions.

### Volume and Market Risk

The periods of elevated trading volume observed in **2021** and **2024–2025** broadly coincide with episodes of higher market volatility identified in the previous notebook. This pattern is consistent with financial market theory, where increased information arrival stimulates trading activity and contributes to larger price fluctuations.

From an economic perspective, fluctuations in trading volume reflect changes in market participation and the speed at which new information is incorporated into asset prices. The sustained increase in trading activity during **2024–2025** suggests stronger investor engagement and a more active price discovery process relative to the quieter market conditions observed during **2022–2023**.

In quantitative finance, trading volume serves as a proxy for market liquidity and information flow. The persistence of volume spikes throughout the sample indicates that liquidity conditions vary over time, supporting the inclusion of volume-based features in predictive models, alpha generation, and market regime identification.
