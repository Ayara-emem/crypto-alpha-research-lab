## Research Question

What are the structural characteristics of the cryptocurrency dataset used throughout this research project?

## Why This Matters

Before conducting any statistical or financial analysis, researchers must understand the scope, completeness, and quality of the available data. Dataset diagnostics help identify issues such as missing observations, inconsistent data types, duplicate timestamps, or limited historical coverage that could bias later analyses.

The purpose of this notebook is to verify that CARL's validated data pipeline provides a reliable foundation for quantitative research.

```python
import pandas as pd
import matplotlib.pyplot as plt

from crypto_alpha_lab.data.storage import load_prices
```

```python
# ==========================================
# Research Configuration
# ==========================================

TICKER = "BTC-USD"

START_DATE = "2020-01-01"
END_DATE = "2026-06-20"

INTERVAL = "1d"

AUTO_ADJUST = True

REFRESH_CACHE = True
```

```python
btc = load_prices(
    ticker=TICKER,
    start=START_DATE,
    end=END_DATE,
    interval=INTERVAL,
    auto_adjust=AUTO_ADJUST,
    refresh=REFRESH_CACHE,
)
```

## Variable Description

| Variable | Description |
|----------|-------------|
| Open | Opening market price for the trading day |
| High | Highest transaction price recorded during the day |
| Low | Lowest transaction price recorded during the day |
| Close | Final market price for the trading day |
| Volume | Number (or value) of units traded during the day |

# Research Conclusion

The dataset satisfies the quality requirements for quantitative financial research. The observations are chronologically ordered, free of duplicate timestamps, and contain no missing values. The sample spans multiple years of Bitcoin trading activity and provides sufficient historical coverage for exploratory analysis, volatility estimation, factor construction, and predictive modeling.
